```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import log_loss

%matplotlib inline

plt.rcParams['font.family'] = 'AppleGothic'
plt.rcParams['axes.unicode_minus'] = False

# 실제 정답 (1=생존, 0=사망)
y_true = np.array([1, 0, 1, 1, 0])
# 모델이 예측한 "생존(1)일 확률"
y_prob = np.array([0.9, 0.1, 0.8, 0.4, 0.3])

# 수동 계산 (공식 그대로)
manual = -np.mean(
    y_true * np.log(y_prob) + (1 - y_true) * np.log(1 - y_prob)
)
print(f"수동 계산 log loss = {manual:.4f}")
print(f"sklearn  log loss = {log_loss(y_true, y_prob):.4f}")   # 동일

# '확신하고 틀린' 경우가 손실을 얼마나 키우는지 비교
print("\n정답=1 일 때 예측 확률별 손실:")
for p in [0.99, 0.9, 0.5, 0.1, 0.01]:
    print(f"  p={p:<4} → 손실 {-np.log(p):.3f}")
```

```python
# sigmoid 함수 정의
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# -10부터 10까지 z값 생성
z = np.linspace(-10, 10, 200)
y = sigmoid(z)

# 그래프
plt.figure(figsize=(8, 5))
plt.plot(z, y, color="steelblue", linewidth=2)
plt.axhline(0.5, color="red", linestyle="--", linewidth=1)    # 0.5 기준선
plt.axvline(0, color="gray", linestyle="--", linewidth=1)     # z=0 기준선
plt.title("Sigmoid Function")
plt.xlabel("z")
plt.ylabel("sigmoid(z)")
plt.grid(alpha=0.3)
plt.show()

# 대표값 확인
for z_val in [-5, -1, 0, 1, 5]:
    print(f"sigmoid({z_val:>2}) = {sigmoid(z_val):.4f}")
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

ranges = [(-2, 2), (-5, 5), (-10, 10), (-100, 100)]

plt.figure(figsize=(12, 8))
for i, (lo, hi) in enumerate(ranges, 1):
    z = np.linspace(lo, hi, 200)
    plt.subplot(2, 2, i)
    plt.plot(z, sigmoid(z), color="steelblue", linewidth=2)
    plt.axhline(0.5, color="red", linestyle="--", linewidth=0.8)
    plt.title(f"z 범위: {lo} ~ {hi}")
    plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
